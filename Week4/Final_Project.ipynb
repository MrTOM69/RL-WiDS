## 2. Download Market Data 

```python
# Choose a stock symbol
symbol = "AAPL"

# Download historical stock price data
data = yf.download(symbol, start="2018-01-01", end="2023-01-01")

# Extract ONLY the closing prices and flatten
prices = data["Close"].values.flatten()

print("Trading days:", len(prices))
print("Sample price:", prices[0], type(prices[0]))
```

## 3. Trading Environment

```python
class TradingEnv:
    def __init__(self, prices):
        self.prices = prices
        self.reset()

    def reset(self):
        self.t = 0
        self.cash = 10000
        self.stock = 0
        self.done = False
        return self._get_state()

    def _get_state(self):
        return np.array([
            self.prices[self.t],
            self.stock
        ], dtype=np.float32)

    def step(self, action):
        price = self.prices[self.t]

        # Buy
        if action == 1 and self.cash >= price and self.stock == 0:
            self.cash -= price
            self.stock = 1

        # Sell
        elif action == 2 and self.stock == 1:
            self.cash += price
            self.stock = 0

        # Move forward
        self.t += 1

        # Termination
        if self.t >= len(self.prices) - 1:
            self.done = True

        # Portfolio value as reward
        reward = self.cash + self.stock * self.prices[self.t]

        return self._get_state(), reward, self.done
```

## 4. Q-Learning Setup

```python
# Q-table: time step × action
Q = np.zeros((len(prices), 3))

alpha = 0.1      # learning rate
gamma = 0.95     # discount factor
epsilon = 0.1    # exploration rate
```

## 5. Train the Agent

```python
env = TradingEnv(prices)
episodes = 100

for episode in range(episodes):
    state = env.reset()

    while not env.done:
        t = env.t

        # Epsilon-greedy
        if np.random.rand() < epsilon:
            action = np.random.choice([0, 1, 2])
        else:
            action = np.argmax(Q[t])

        _, reward, _ = env.step(action)

        # Bellman update
        Q[t, action] += alpha * (
            reward +
            gamma * np.max(Q[min(t + 1, len(prices) - 1)]) -
            Q[t, action]
        )

print("Training completed.")
```

## 6. Evaluate Trained Agent

```python
env = TradingEnv(prices)

while not env.done:
    t = env.t
    action = np.argmax(Q[t])
    env.step(action)

final_value = env.cash + env.stock * prices[-1]
print("Final portfolio value (RL Agent):", final_value)
```
